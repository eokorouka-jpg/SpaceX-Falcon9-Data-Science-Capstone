```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

## Railway and highway proximity analysis
### We'll use CCAFS LC-40 as our reference launch site.

```python
# Prepare the reference coordinates

launch_site = (28.562302, -80.577356)

print("Reference launch site: CCAFS LC-40")
print("Latitude:", launch_site[0])
print("Longitude:", launch_site[1])
```

```python
# Find railway and highway locations

import requests

lat, lon = 28.562302, -80.577356

query = f"""
[out:json][timeout:60];
(
  way["railway"="rail"](around:20000,{lat},{lon});
  way["highway"~"^(motorway|trunk|primary|secondary|tertiary)$"]
      (around:20000,{lat},{lon});
);
out center;
"""

url = "https://overpass.private.coffee/api/interpreter"

try:
    response = requests.post(
        url,
        data={"data": query},
        timeout=90
    )
    response.raise_for_status()

    features = response.json()["elements"]

    for feature in features[:15]:
        tags = feature.get("tags", {})
        center = feature.get("center", {})

        print(
            tags.get("name", "Unnamed"),
            "| Type:", tags.get("railway", tags.get("highway")),
            "| Latitude:", center.get("lat"),
            "| Longitude:", center.get("lon")
        )

    print("\nTotal features found:", len(features))

except requests.RequestException as error:
    print("Map data request failed:", error)
```

```python
from math import radians, sin, cos, sqrt, atan2

launch_site = (28.562302, -80.577356)
railway_point = (28.572277, -80.653919)

def calculate_distance(point1, point2):
    R = 6371  # Earth's radius in kilometres

    lat1, lon1 = map(radians, point1)
    lat2, lon2 = map(radians, point2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        sin(dlat / 2)**2
        + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    )

    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    return R * c

railway_distance = calculate_distance(
    launch_site, railway_point
)

print(f"Distance to railway: {railway_distance:.2f} km")
```
