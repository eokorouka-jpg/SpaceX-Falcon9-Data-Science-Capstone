# SpaceX Launch Records Dashboard Analysis

## Project Overview

This project uses Python, Pandas, Plotly and Dash to build an interactive dashboard for analysing SpaceX Falcon 9 first-stage landing outcomes.

The dashboard allows users to explore landing success rates across launch sites, payload mass ranges and booster versions.

## Key Findings

### 1. Launch Site with the Highest Number of Successful Landings

KSC LC-39A recorded the highest number of successful landings, with **10 successful landings out of 13 launches**.

### 2. Launch Site with the Highest Landing Success Rate

KSC LC-39A achieved the highest landing success rate of **76.92%**.

### 3. Payload Range with the Highest Landing Success Rate

Payloads between **2,000 and 4,000 kg** recorded the highest observed landing success rate of **61.90%**.

### 4. Payload Range with the Lowest Landing Success Rate

Payloads between **6,000 and 8,000 kg** recorded a **0% landing success rate** in the analysed dataset.

### 5. Booster Version with the Highest Landing Success Rate

Falcon 9 Block 5 (B5) achieved the highest observed landing success rate of **100%**.

Other booster success rates included:

- Full Thrust (FT): 66.67%
- Block 4 (B4): 54.55%
- Falcon 9 v1.1: 6.67%
- Falcon 9 v1.0: 0%

## Conclusion

The SpaceX dashboard analysis demonstrates that landing outcomes vary across launch sites, payload mass ranges and booster versions.

KSC LC-39A demonstrated the strongest landing performance among the analysed launch sites, while Falcon 9 Block 5 achieved the highest observed success rate among booster versions.

These findings suggest that launch-site characteristics, payload mass and booster technology are relevant factors when examining SpaceX first-stage landing outcomes.

The results are based on the historical dataset used in the IBM Data Science Capstone project and should not be interpreted as predictions of future landing outcomes.

```python
import pandas as pd

df = pd.read_csv("spacex_launch_dash.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

## Landing Success Analysis by Launch Site

```python
# Calculate landing success by launch site
site_results = df.groupby("Launch Site")["class"].agg(
    successful="sum",
    total="count",
    success_rate="mean"
)

# Convert success rate to percentage
site_results["success_rate"] = (
    site_results["success_rate"] * 100
)

# Display results
display(site_results.round(2))
```

## Landing Success Analysis by Payload Mass

```python
# Group payload masses into ranges
df["Payload Range"] = pd.cut(
    df["Payload Mass (kg)"],
    bins=[0, 2000, 4000, 6000, 8000, 10000],
    include_lowest=True
)

# Calculate landing success for each payload range
payload_results = df.groupby(
    "Payload Range", observed=True
)["class"].agg(
    successful="sum",
    total="count",
    success_rate="mean"
)

# Convert success rate to percentage
payload_results["success_rate"] *= 100

# Display results
display(payload_results.round(2))
```

## Landing Success Analysis by Booster Version

```python
# Calculate landing success by booster version
booster_results = df.groupby(
    "Booster Version Category"
)["class"].agg(
    successful="sum",
    total="count",
    success_rate="mean"
)

# Convert success rate to percentage
booster_results["success_rate"] *= 100

# Sort booster versions by success rate
booster_results = booster_results.sort_values(
    by="success_rate",
    ascending=False
)

# Display results
display(booster_results.round(2))
```
